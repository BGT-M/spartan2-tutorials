```python
# Import spartan and dependencies
import spartan as st
```

```python
# Import numpy and sparse
import numpy as np
import sparse
```

# CPU Dense

```python
# Create a 3x4 dense tensor from numpy array
A = st.DTensor.from_numpy(np.random.rand(3, 4))
print(A)
```

## Get Attributes

```python
# Access shape, dtype, and transpose
print(A.shape)
print(A.dtype)
print(A.T)
```

## Index and Slice

```python
# Indexing and slicing
print(len(A))
print(A[0])
print(A[:, 1])
print(A[:, 0:2])
```

## Reduction operations

```python
# Reduction: sum over all elements or along an axis (method API)
print(A.sum())
print(A.sum(axis=0))
```

```python
# Same reductions via spartan functional API
print(st.sum(A))
print(st.sum(A, axis=0))
```

## Binary operations

```python
# Create second dense tensor for binary operations
B = st.DTensor.from_numpy(np.random.rand(3, 4))
```

```python
# Element-wise addition
print(A+B)
print(st.add(A, B))
```

```python
# Matrix multiplication
print(A.dot(B.T))
print(st.dot(A, B.T))
```

# CPU Sparse

```python
# Create a sparse tensor (STensor) from a sparse numpy array
A = np.random.rand(3, 4)
A[A<0.8] = 0
A = st.STensor.from_numpy(A)
print(A)
print(A.todense())
```

## Get Attributes

```python
# Attributes of sparse tensor
print(A.shape)
print(A.dtype)
print(A.T)
```

## Index and Slice

```python
# Indexing and slicing on sparse tensor
print(len(A))
print(A[0])
print(A[:, 1])
print(A[:, 0:2])
```

## Reduction operations

```python
# Reduction on sparse tensor (method API)
print(A.sum())
print(A.sum(axis=0).todense())
```

```python
# Same reductions via spartan functional API
print(st.sum(A))
print(st.sum(A, axis=0).todense())
```

## Binary operations

```python
# Create second sparse tensor
B = np.random.rand(3, 4)
B[B<0.8] = 0
B = st.STensor.from_numpy(B)
```

```python
# Element-wise addition on sparse tensors
print(A+B)
print((A+B).todense())
print(st.add(A, B))
print(st.add(A, B).todense())
```

```python
# Matrix multiplication on sparse tensors
print(A.dot(B.T))
print(A.dot(B.T).todense())
print(st.dot(A, B.T))
print(st.dot(A, B.T).todense())
```
